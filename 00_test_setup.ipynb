# Test notebook: is everything ready?

**Do this before the tutorial.** It takes one minute and tells you whether your laptop (or Colab) can run the three tutorial notebooks.

**What to do:** click on the grey code cell below and press **Shift+Enter** (hold Shift, press Enter).

**What you should see:** the last line of the output is the single word `ready`.

The first code cell of every notebook in this tutorial does the same job of finding the data, so this also tests that step.

```python
import os, sys, subprocess
URL = "https://github.com/harish-srini/cocogel_tutorial.git"   # where the materials live
REPO = "cocogel_tutorial"                            # the folder that git will create
if "google.colab" in sys.modules and not os.path.exists("data"):   # on Colab, data not here yet
    if not os.path.exists(REPO):                     # not downloaded yet
        subprocess.run(["git", "clone", "--depth", "1", URL], check=True)
    os.chdir(REPO)                                   # step into the downloaded folder
assert os.path.exists("data/bsa_dls.csv"), "The data folder was not found. See the table below."
import numpy, scipy, matplotlib, pandas              # stops here with an error if one is missing
print("ready")
```

## What just happened?

| Code | Meaning |
|---|---|
| `import os, sys, subprocess` | `import` loads a library, a collection of ready-made tools. These three come with Python. `os` talks to the operating system, `sys` describes the running Python, and `subprocess` starts other programs. |
| `if "google.colab" in sys.modules and not os.path.exists("data"):` | This is `True` only on Google Colab, and only while the data aren't here yet. The indented lines below it run only then. |
| `subprocess.run(["git", "clone", ...], check=True)` | On Colab the notebook arrives without its data, so this copies the whole folder from GitHub. |
| `os.chdir(REPO)` | Moves into the copied folder. |
| `assert ...` | Checks that a data file can be found, and stops with the message if not. |
| `import numpy, scipy, matplotlib, pandas` | Loads the four libraries we use: numbers, fitting, plots and tables. If one is missing, Python stops here with an error. |
| `print("ready")` | Writes the word `ready`. Python reaches this line only if everything above worked. |

## If you did not see `ready`

| Message | Cause | Fix |
|---|---|---|
| `ModuleNotFoundError: No module named 'numpy'` (or `scipy`, `matplotlib`, `pandas`) | The package is not installed in the Python that JupyterLab is using. | Close JupyterLab. Activate the environment, because your prompt should start with `(cocogel)`. Run the install line from the README again, then start `jupyter lab` again. |
| `AssertionError: The data folder was not found` | This notebook is not next to the folder `data`. | Extract the ZIP if you have not done so. In the terminal use `cd` to go to the folder holding this notebook, then start `jupyter lab` again. |
| `CalledProcessError` on Colab | `git` could not reach GitHub. | Check your internet connection, then choose Runtime, Restart session, and run the cell again. |
| Nothing happens and `[*]` stays on the left | Python is still starting. | Wait a minute. If nothing changes, use the menu Kernel, then Restart Kernel. |

If it still fails, don't lose time: use Google Colab, as described in the README. It runs in a browser and needs no installation.
