## The physics, in words

In a crowded solution each protein is hemmed in by its neighbours. For a short time it moves in the free space around it, almost as if it were alone. To go further, its neighbours must first make room, and that is slower. So its mean squared displacement (MSD) bends, from one straight line on a log-log plot to a lower one, and in between it grows more slowly than $\tau$. As the lecture showed, $f$ follows the MSD, so the decay is no longer a single exponential. It is **stretched**: faster than an exponential at the start and slower at the end.

Crowding also creates **structure**: neighbours keep a preferred distance. A density ripple whose wavelength matches this distance is the pattern the particles prefer to be in, so it costs little, and when a fluctuation is cheap to make it is slow to decay. So at the peak of the structure factor, $D(q)$ has a dip. Pierre-Gilles de Gennes pointed this out in 1959, for neutrons on liquids. It is called **de Gennes narrowing**.

## The physics, in equations

**The model.** A stretched exponential for the density correlation, with exponent $\alpha$:

$$f(q,\tau) = \exp\!\left[-(\tilde\Gamma\tau)^\alpha\right], \qquad g_2(q,\tau) = 1 + \beta(q)\exp\!\left[-2(\tilde\Gamma\tau)^\alpha\right].$$

The 2 is there because the detector records intensity, and $g_2-1$ goes as $f^2$. For $\alpha = 1$ this is the exponential of notebook 01.

**Why this form?** For Gaussian displacements, $f = \exp[-q^2\langle\Delta r^2(\tau)\rangle/6]$. A straight MSD, $\langle\Delta r^2\rangle = 6D\tau$, gives one exponential. An MSD that goes as $\tau^\alpha$, with $\alpha<1$, gives exactly the stretched form. For interacting particles this link is only a guide, because the collective $f$ is not just this formula.

**The mean rate.** Define the mean relaxation time as the area under $f$:

$$\frac{1}{\Gamma} = \int_0^\infty f(q,\tau)\,d\tau = \frac{\Gamma_f(1/\alpha)}{\alpha\,\tilde\Gamma} \qquad\Rightarrow\qquad \Gamma = \tilde\Gamma\,\frac{\alpha}{\Gamma_f(1/\alpha)} .$$

Here $\Gamma_f$ is Euler's **gamma function**, $\Gamma_f(z) = \int_0^\infty x^{z-1}e^{-x}dx$. The subscript $f$ keeps it apart from the rate $\Gamma$: the same Greek letter means two different things in the literature. For $\alpha=1$, $\Gamma_f(1)=1$, and so $\Gamma = \tilde\Gamma$, as it must be.

**The test.** Divide by $q^2$:

$$D(q) = \frac{\Gamma(q)}{q^2}.$$

If $\Gamma=Dq^2$ held, $D(q)$ would be flat. Near the structure peak, at short times, theory gives $D(q) = D_0\,H(q)/S(q)$, where $S(q)$ is the structure factor and $H(q)$ the hydrodynamic function. Where $S(q)$ peaks, $D(q)$ dips.

# Part C. Fitting a stretched decay

## Step 6. Write the model

**What this cell does and why.** The function takes $\tau$ and three parameters. **The first input must be the x variable** for `curve_fit`.

We will keep $\alpha$ **fixed** for each sample. Why not fit it? Because at a single $q$, $\alpha$ and $\tilde\Gamma$ trade off against each other: a change in one can be mostly undone by a change in the other. Within the noise here, an independent $\alpha$ at each $q$ scatters widely. The values come from the fit of the paper to all $q$ together. You can test this in the Bonus part.

**`np.exp(x)`** is $e^x$, element by element, and `**` raises to a power.

**Units.** We use $\tau$ in μs, so $\tilde\Gamma$ is in μs⁻¹, and the combination $\tilde\Gamma\tau$ has no unit, as a quantity raised to a power must.

````python
def g2_stretched(tau, beta, Gamma_tilde, alpha):
    """g2 = 1 + beta exp[-2 (Gamma_tilde tau)^alpha]. tau in us, Gamma_tilde in 1/us."""
    return 1 + beta * np.exp(-2 * (Gamma_tilde * tau)**alpha)
print(g2_stretched(0.0, 0.05, 1.0, 0.9), g2_stretched(100.0, 0.05, 1.0, 0.9))    # expect 1.05 and 1.0
````

**What you should see.** `1.05` and a value very close to 1. At $\tau=0$ the function gives $1+\beta$, and at a long lag it gives 1.

## Step 7. Fit one $q$ with $\alpha$ fixed, and convert to the mean rate

**What this cell does and why.** We fit one curve and then apply $\Gamma = \tilde\Gamma\,\alpha/\Gamma_f(1/\alpha)$.

**`curve_fit` takes a function whose free inputs are all to be fitted.** So we hand over a small one with $\alpha$ already set:

`lambda t, b, G: g2_stretched(t, b, G, alpha)`

A **lambda** is a one-line function without a name. This one takes `t`, `b` and `G`, and calls our model with the fixed `alpha`.

**The arguments.** As in the earlier notebooks:

| Argument | Meaning |
|---|---|
| the function | the model with the free parameters |
| `tau`, `g2` | the data |
| `p0=[g2[0] - 1, 0.2]` | starting guess: $\beta$ from the first point, and $\tilde\Gamma$ = 0.2 μs⁻¹, the right order of magnitude |
| `sigma=err`, `absolute_sigma=True` | the errors, as true sizes |
| `bounds=([0, 0], [1, np.inf])` | allowed ranges, as `([lower], [upper])`: $0<\beta<1$, and $\tilde\Gamma>0$. A negative value inside a fractional power would have no meaning. |

**It gives back** `popt`, the best parameters, and `pcov`, the covariance matrix. **`np.sqrt(np.diag(pcov))`** gives the uncertainties. `np.diag` picks out the diagonal.

````python
c, q = 400, 0.375                                   # <-- CHANGE ME   sample and q
alpha = ALPHA[c]
d = data[c][np.isclose(data[c]["q_nm-1"], q)]
tau, g2, err = d["lag_time_s"].values * 1e6, d["g2"].values, d["g2_err"].values     # tau in us
popt, pcov = curve_fit(lambda t, b, G: g2_stretched(t, b, G, alpha), tau, g2, p0=[g2[0] - 1, 0.2],
                       sigma=err, absolute_sigma=True, bounds=([0, 0], [1, np.inf]))
Gamma_mean = popt[1] * alpha / gamma_function(1 / alpha)
print(f"beta = {popt[0]:.4f},  Gamma_tilde = {popt[1]:.4f}/us,  mean rate = {Gamma_mean:.4f}/us")
````

**What you should see, at 400 mg/ml and $q=0.375$.** $\beta \approx 0.046$, $\tilde\Gamma \approx 0.092\ \mathrm{\mu s^{-1}}$ and a mean rate of about $0.085\ \mathrm{\mu s^{-1}}$.

**How to read it.** With $\alpha=0.86$, the factor $\alpha/\Gamma_f(1/\alpha)$ is 0.93, a correction of 7 %. So in this sample, ignoring it would give a systematic error larger than the statistical one.

**Common errors.** `ValueError: x0 is infeasible`: the guess lies outside the bounds; this happens if `g2[0] - 1` is negative because of noise. `IndexError` or an empty selection: `q` is not in the file, for example $q=0.525$ at 70 mg/ml.

## Step 8. Show the fit, with a single exponential for comparison

**What this cell does and why.** Does the fit follow the data? And can we see the stretching? We draw the stretched fit, and a single exponential with the **same mean rate**, $\alpha=1$.

**`np.logspace(a, b, 300)`** gives 300 numbers evenly spaced on a log axis, from $10^a$ to $10^b$. **`plt.semilogx`** plots with a logarithmic x axis.

````python
tau_fine = np.logspace(-1, 1.6, 300)              # from 0.1 to about 40 us
plt.errorbar(tau, g2, yerr=err, fmt="o", ms=4, label=f"{c} mg/ml, q = {q}")
plt.semilogx(tau_fine, g2_stretched(tau_fine, popt[0], popt[1], alpha),
             label=fr"fit, $\alpha$ = {alpha}")
plt.semilogx(tau_fine, g2_stretched(tau_fine, popt[0], Gamma_mean, 1.0), "--",
             label=r"same mean rate, $\alpha$ = 1")
plt.xlabel(r"lag time $\tau$ (μs)"); plt.ylabel(r"$g_2(q,\tau)$")
plt.legend(); plt.show()
````

**What you should see.** The fit follows the data. The dashed exponential lies slightly above the fit over most of the decay, and crosses below it only in the far tail. The difference is modest, about the size of the error bars at this noise level.

**How to read it.** The stretched curve drops faster at the start and keeps a longer tail. That is the bending MSD: fast motion in the free space around a protein, then slower motion that needs the neighbours to rearrange. It is the signature of crowding, but at this $\alpha$ you need a careful fit to see it.

# Part D. Is $D(q)$ flat?

## Step 9. Fit every $q$ of both samples

**What this cell does and why.** The same fit as in Step 7, in a loop. We first put the fit into a function, so that the loop stays short and readable.

**`df.groupby("q_nm-1")`** splits the table by $q$. Each round of the loop gives $q$ and its part of the table. **`.values`** gives a plain array from a column. `return a, b, c` gives back three values at once. **`rows.append({...})`** adds a row to a list, as a dictionary of column names and values, and **`pd.DataFrame(rows)`** turns the list into a table.

**The result.** $D(q) = \Gamma/q^2$. Since $\Gamma$ is in μs⁻¹ and $q^2$ in nm⁻², $D$ comes out in nm²/μs. With $\alpha$ fixed, the relative uncertainty of $D$ equals that of $\tilde\Gamma$.

````python
def fit_one_q(d, alpha):
    tau, g2, err = d["lag_time_s"].values * 1e6, d["g2"].values, d["g2_err"].values
    popt, pcov = curve_fit(lambda t, b, G: g2_stretched(t, b, G, alpha), tau, g2, p0=[g2[0] - 1, 0.2],
                           sigma=err, absolute_sigma=True, bounds=([0, 0], [1, np.inf]))
    return popt[0], popt[1], np.sqrt(pcov[1, 1])    # beta, Gamma_tilde, uncertainty of Gamma_tilde
````

````python
rows = []
for c in data:
    for q, d in data[c].groupby("q_nm-1"):
        beta, G_tilde, G_err = fit_one_q(d, ALPHA[c])
        D_q = G_tilde * ALPHA[c] / gamma_function(1 / ALPHA[c]) / q**2       # nm^2/us
        rows.append({"c_mg_ml": c, "q": q, "beta": beta, "D": D_q, "D_err": D_q * G_err / G_tilde})
results = pd.DataFrame(rows)
results
````

**What you should see.** A table of 14 rows. At 70 mg/ml, $D$ falls from about 2.64 to about 2.32 nm²/μs. At 400 mg/ml, $D$ is between about 0.48 and 0.75 nm²/μs.

**A new thing: a loop inside a loop.** The outer loop runs over the samples, and for each sample the inner loop runs over its $q$ values. Note the growing indentation.

**Common errors.** `NameError: name 'gamma_function' is not defined`: you skipped Step 2. `ValueError: x0 is infeasible`: see Step 7.

## Step 10. Plot $D(q)/D_0$, with the positions of the $S(q)$ peaks

**What this cell does and why.** Dividing by $D_0$ gives a number without units: how much slower than a dilute particle? On a logarithmic axis, equal ratios have equal distances, so both samples can be compared on one plot. **`plt.yscale("log")`** sets this, and **`plt.axvline(x)`** draws a vertical line at the $S(q)$ peak of each sample. **`plt.ylim(a, b)`** sets the range of the y axis, **`plt.yticks(positions, labels)`** chooses the numbers on the axis, and **`plt.minorticks_off()`** removes the small unlabelled ticks. **`enumerate(data)`** gives a counter `i` (0, 1, ...) along with each sample, which we use to pick a marker, `"os"[i]` (the letter `o` for circles or `s` for squares), and a colour, `"C0"` or `"C1"`, matplotlib's first two default colours.

````python
for i, c in enumerate(data):                       # enumerate gives a counter i as well
    r = results[results["c_mg_ml"] == c]
    plt.errorbar(r["q"], r["D"] / D0, yerr=r["D_err"] / D0, fmt="os"[i], ms=7, color=f"C{i}",
                 label=f"{c} mg/ml")
    plt.axvline(S_PEAK[c], color=f"C{i}", ls="--")
plt.axhline(1, color="0.4", ls=":"); plt.yscale("log"); plt.ylim(0.1, 1.4)
plt.yticks([0.1, 0.2, 0.5, 1], ["0.1", "0.2", "0.5", "1"]); plt.minorticks_off()
plt.xlabel(r"$q$ (nm$^{-1}$)"); plt.ylabel(r"$D(q)/D_0$"); plt.legend(); plt.show()
````

**What you should see.** The dotted line at 1 is the dilute limit. The blue circles at 70 mg/ml lie a little below it, from about 0.87 down to about 0.76. The orange squares at 400 mg/ml lie far below it, from about 0.24 down to about 0.16, and then rise again. The dashed vertical lines mark the $S(q)$ peaks at 0.36 and 0.50 nm⁻¹.

**How to read it: this is the result.**

- **$D(q)$ is not flat.** $\Gamma = Dq^2$ with one number $D$ fails. The curve has a shape, and the shape follows the structure.
- **400 mg/ml: a clear dip** near the $S(q)$ peak at 0.50 nm⁻¹, four to six times slower than a dilute particle. $D$ rises again clearly at larger $q$.
- **70 mg/ml: a gentle fall.** Be careful with this one. The file stops at $q = 0.425\ \mathrm{nm^{-1}}$, so we see only the left side of any dip. It falls from 0.87 towards 0.76 around the $S(q)$ peak at 0.36 nm⁻¹, and the last points agree with each other within their errors. So this file by itself shows a fall towards a minimum, not a full dip.
- Even at 70 mg/ml, the sample is **already below 1**: it feels its neighbours.

## Step 11. Where is the dip?

**What this cell does and why.** We read off the minimum of $D$ in each sample, and compare it with the peak of $S(q)$. **`.idxmin()`** gives the row label of the smallest value, and **`.loc[label]`** picks out that row. (Compare `.iloc`, which uses the position.)

````python
for c in data:
    r = results[results["c_mg_ml"] == c]
    m = r.loc[r["D"].idxmin()]
    print(f"{c} mg/ml: smallest D = {m['D']:.3f} +- {m['D_err']:.3f} nm^2/us at q = {m['q']:.3f} per nm")
    print(f"   S(q) peak at {S_PEAK[c]} per nm")
r70, r400 = results[results["c_mg_ml"] == 70], results[results["c_mg_ml"] == 400]
ratio = r70[np.isclose(r70["q"], 0.375)]["D"].iloc[0] / r400[np.isclose(r400["q"], 0.375)]["D"].iloc[0]
print(f"D at q = 0.375: 70 mg/ml is {ratio:.1f} times faster than 400 mg/ml")
````

**What you should see.**

- At 70 mg/ml, the smallest $D$ is about $2.32 \pm 0.15$ nm²/μs at $q = 0.375\ \mathrm{nm^{-1}}$, close to the $S(q)$ peak at 0.36.
- At 400 mg/ml, the smallest $D$ is about $0.477 \pm 0.017$ nm²/μs at $q = 0.525\ \mathrm{nm^{-1}}$, close to the $S(q)$ peak at 0.50. Note that $q=0.425$ gives almost the same value, $0.496 \pm 0.012$, so the position of the minimum isn't sharply defined by one point. The dip is broad, and its position agrees with the peak within the spacing of the $q$ grid.
- At $q = 0.375$, the 70 mg/ml sample is about 3.8 times faster.

**How to read it.** The dynamics are slowest at the $q$ of the most favoured density pattern. That is **de Gennes narrowing**. "Narrowing" refers to the frequency spectrum: a slower decay means a narrower spectrum. And it can be seen only at the molecular length scales that X-rays reach. At the $q$ values of DLS, around 0.02 nm⁻¹, we are far to the left of this plot, and this dip is out of sight.

**Question 3.** Why does $D(q)$ rise again at $q$ above the peak? And what would you expect at very large $q$?

<details>
<summary>Answer</summary>

Beyond the peak, S(q) falls, so the fluctuations are less favoured and relax faster. At very large q, S(q) → 1 and you probe distances smaller than the gap between neighbours, so the particle sees mostly its own free motion over those short distances. D(q) tends towards a short-time self-diffusion coefficient. It is still lower than D₀, because hydrodynamic interactions with neighbours remain.

</details>

No. β only sets the height of the decay, not its rate. But it makes the measurement much harder: the signal over the noise is much smaller, as the larger error bars show.
