## The physics, in words

Shine a laser on a dilute protein solution. The proteins scatter the light, and the scattered waves interfere. On a detector you see a pattern of bright and dark spots, called **speckle**. The proteins move, so the speckle flickers.

A small protein diffuses quickly, so the speckle flickers quickly. A large protein diffuses slowly, so the speckle flickers slowly. The speed of the flicker therefore carries the size of the protein.

We can't follow a flicker of some tens of microseconds by eye, so we let a correlator do it. It asks: if the intensity is high now, is it still high a lag time $\tau$ later? For short $\tau$ the answer is yes, and for long $\tau$ the memory is gone. How fast that memory is lost is the decay rate $\Gamma$.

The detector angle selects a length scale. A larger angle looks at finer density ripples, and finer ripples are erased faster by diffusion. This is the test we will use: changing the angle must change the rate in one very specific way.

## The physics, in equations

**1. What the detector angle selects.** The scattering vector is

$$q = \frac{4\pi n}{\lambda}\,\sin\!\left(\frac{\theta}{2}\right).$$

The density ripple that we observe has a wavelength $2\pi/q$.

**2. How one ripple decays.** Diffusion erases a ripple of wave vector $q$ exponentially. In the lecture's notation, the normalised density correlation is

$$f(q,\tau) = \exp(-\Gamma\tau), \qquad \Gamma = Dq^2 .$$

Many books call $f$ the field correlation function $g_1$. We will use $f$.

**3. What the detector records.** We record intensity, not density. For speckle (the Siegert relation),

$$g_2(q,\tau) = \frac{\langle I(t)\,I(t+\tau)\rangle}{\langle I\rangle^2} = 1 + \beta\,|f(q,\tau)|^2 .$$

**4. The model we fit.** Put 2 into 3:

$$g_2(q,\tau) = 1 + \beta\exp(-2\Gamma\tau).$$

**5. From the slope to the size.** Stokes–Einstein connects $D$ with the radius of a sphere:

$$R_h = \frac{k_BT}{6\pi\eta D}.$$

# Part C. The decay rate at one angle

## Why the 2 in the exponent?

The exponent has $2\Gamma$, and the factor of 2 catches everyone once, so let us be explicit.

- $\Gamma$ is the decay rate of the **density** correlation $f = e^{-\Gamma\tau}$. That is the quantity with the physics in it: $\Gamma = Dq^2$.
- The detector records **intensity**, and intensity is the square of the field. For speckle, the Siegert relation says that $g_2 - 1$ is proportional to $|f|^2$.
- Squaring an exponential doubles its exponent: $\left(e^{-\Gamma\tau}\right)^2 = e^{-2\Gamma\tau}$.

So $g_2 - 1$ decays **twice as fast** as $f$. Fit $g_2$ with $\exp(-\Gamma\tau)$ by mistake, and your "$\Gamma$" is twice too large. You will test this in Question 4.

## Step 8. Write the model as a function

**What this cell does and why.** `curve_fit` needs the model as a Python function. **The first input must be the x variable,** here `tau`. The other inputs are the parameters that `curve_fit` will adjust.

**`np.exp(x)`** is $e^x$. It works element by element, so if `tau` is an array of 136 lag times, the function gives back an array of 136 values. No loop is needed.

Units: `tau` in s and `Gamma` in s⁻¹, so that the exponent has no unit, as it must.

````python
def g2_model(tau, beta, Gamma):
    """g2 = 1 + beta exp(-2 Gamma tau). tau in s, Gamma in 1/s, beta has no unit."""
    return 1 + beta * np.exp(-2 * Gamma * tau)

print(g2_model(0.0, 0.8, 25000.0), g2_model(1.0, 0.8, 25000.0))    # test: tau = 0 and tau very long
````

**What you should see.** `1.8 1.0`. At $\tau=0$ the model gives $1+\beta$, and at a very long lag it gives 1. Testing a function at values where you know the answer is a good habit.

The text in triple quotes is a *docstring*, a note for human readers. Python ignores it.

## Step 9. Prepare one angle and make a starting guess

**What this cell does and why.** A fit improves a guess step by step, so it needs a reasonable start. We make the guess from the data, so that it works at every angle.

- **`.values`** turns a pandas column into a plain NumPy array, which is what `curve_fit` likes.
- **`g2[0]`** is the first element. With $\tau$ close to 0, $g_2 \approx 1+\beta$, so `g2[0] - 1` guesses $\beta$.
- **The guess for $\Gamma$.** The curve has dropped by a factor $e$ when $2\Gamma\tau = 1$. So we find the lag $\tau_e$ where $g_2 - 1$ is closest to $\beta/e$, and then $\Gamma \approx 1/(2\tau_e)$.
- **`np.abs(x)`** is the absolute value, element by element, and **`np.argmin(x)`** gives the **position** of the smallest element. Together they find the point closest to a target. **`np.e`** is the number $e$.

````python
angle = 90                                     # <-- CHANGE ME   any angle from the file
d = df[df["angle_deg"] == angle]
tau, g2, err = d["lag_time_s"].values, d["g2"].values, d["g2_err"].values
beta_guess = g2[0] - 1
i_e = np.argmin(np.abs(g2 - 1 - beta_guess / np.e))    # position of the point nearest to beta/e
Gamma_guess = 1 / (2 * tau[i_e])
print(f"Starting guess: beta = {beta_guess:.2f}, Gamma = {Gamma_guess:.0f} per second")
````

**What you should see.** About $\beta \approx 0.79$ and $\Gamma \approx 2.4\times10^{4}\ \mathrm{s^{-1}}$ at 90°. That is already close to the lecture's value, from one line of arithmetic.

**Common error.** `KeyError: 'g2 '` means that the column name is misspelled. Here it has an extra space.

## Step 10. Fit

**What this cell does and why.** `curve_fit` finds the $\beta$ and $\Gamma$ that make the model pass as close as possible to the data, by minimising

$$\chi^2 = \sum_i \left(\frac{g_{2,i} - \text{model}(\tau_i)}{\sigma_i}\right)^2 .$$

Each point counts according to its error $\sigma_i$: precise points count more.

**The arguments of `curve_fit(g2_model, tau, g2, p0=..., sigma=..., absolute_sigma=True, bounds=...)`:**

| Argument | Meaning |
|---|---|
| `g2_model` | the model function. Note: no round brackets, because we hand over the function itself. |
| `tau`, `g2` | the x and y data |
| `p0` | "parameters at step zero", the starting guess, in the order of the function's inputs: `[beta, Gamma]` |
| `sigma` | the error of each point, here `g2_err` |
| `absolute_sigma=True` | "these errors are true sizes, not only relative weights". This affects the uncertainty that comes back. |
| `bounds` | the allowed range, given as `([lower values], [upper values])`. Here $0<\beta<2$ and $\Gamma>0$: a negative rate or contrast has no meaning. `np.inf` is infinity, so "no upper limit". |

**What comes back: two things,** which we store with `popt, pcov = ...`:

- **`popt`**, the optimal parameters: `popt[0]` is $\beta$ and `popt[1]` is $\Gamma$.
- **`pcov`**, the covariance matrix, a 2 × 2 table. Its diagonal holds the squared uncertainties. **`np.diag(pcov)`** picks out the diagonal, and **`np.sqrt`** takes the square root, so `np.sqrt(np.diag(pcov))` gives the uncertainties.

**Reduced $\chi^2$.** Divide $\chi^2$ by the number of data points minus the number of parameters. If the model is right and the errors are right, this is close to 1. `np.sum` adds up all elements of an array.

````python
popt, pcov = curve_fit(g2_model, tau, g2, p0=[beta_guess, Gamma_guess], sigma=err,
                       absolute_sigma=True, bounds=([0, 0], [2, np.inf]))
beta_err, Gamma_err = np.sqrt(np.diag(pcov))       # uncertainties
residuals = g2 - g2_model(tau, *popt)              # data minus model
chi2_red = np.sum((residuals / err)**2) / (len(tau) - 2)
print(f"beta  = {popt[0]:.4f} +- {beta_err:.4f}")
print(f"Gamma = {popt[1]:.0f} +- {Gamma_err:.0f} per second")
print(f"reduced chi-squared = {chi2_red:.2f}")
````

**What you should see, at 90°.** $\beta \approx 0.798$, $\Gamma \approx 25\,240 \pm 50\ \mathrm{s^{-1}}$ and a reduced $\chi^2 \approx 0.9$.

**What it means.** This agrees with the lecture's $\Gamma \approx 25\,200\ \mathrm{s^{-1}}$. The reduced $\chi^2$ close to 1 says that one exponential describes this curve within the errors.

**About `*popt`.** The star unpacks the array, so `g2_model(tau, *popt)` means `g2_model(tau, popt[0], popt[1])`.

**Common errors**

| Message | Cause | Fix |
|---|---|---|
| `RuntimeError: Optimal parameters not found` | The fit did not converge, usually from a poor guess. | Check `p0`: is it near the plot? Check that the data are the ones you expect. |
| `ValueError: x0 is infeasible` | The starting guess lies outside the `bounds`. | Make the guess positive, or widen the bounds. |
| `TypeError: g2_model() takes 3 positional arguments...` | The number of values in `p0` does not match the function's inputs. | Give one value in `p0` for each parameter. |

# Part D. All nine angles

## Step 12. Fit in a loop

**What this cell does and why.** We repeat the fit of Steps 9 and 10 at each angle, and collect the results in a new table. The physics is the same, and only the bookkeeping is new.

| Command | What it does |
|---|---|
| `rows = []` | An empty list, to be filled. |
| `for angle, d in df.groupby("angle_deg"):` | **`groupby`** splits the table by the values of a column. Each round of the loop gives the angle and the part `d` of the table at that angle. |
| `rows.append({...})` | Adds one element to the end of the list. The element is a *dictionary*: pairs of name and value in curly brackets. |
| `pcov[1, 1]` | The element in row 1, column 1 of the matrix, which is the squared uncertainty of $\Gamma$. |
| `return a, b, c` | A function can give back several values at once. `beta, Gamma, Gamma_err = fit_one_angle(d)` stores them under three names. |
| `pd.DataFrame(rows)` | Makes a table from the list of dictionaries, one row for each. |

````python
def fit_one_angle(d):
    tau, g2, err = d["lag_time_s"].values, d["g2"].values, d["g2_err"].values
    guess = [g2[0] - 1, 1 / (2 * tau[np.argmin(np.abs(g2 - 1 - (g2[0] - 1) / np.e))])]
    popt, pcov = curve_fit(g2_model, tau, g2, p0=guess, sigma=err,
                           absolute_sigma=True, bounds=([0, 0], [2, np.inf]))
    return popt[0], popt[1], np.sqrt(pcov[1, 1])       # beta, Gamma, uncertainty of Gamma
````

**What you should see.** Nothing is printed: this cell only defines the function. It is Steps 9 and 10 packed into one recipe, which takes the part of the table at one angle and returns three numbers.

**Now the loop.**

````python
rows = []
for angle, d in df.groupby("angle_deg"):
    beta, Gamma, Gamma_err = fit_one_angle(d)
    rows.append({"angle_deg": angle, "q": d["q_nm-1"].iloc[0], "beta": beta,
                 "Gamma": Gamma, "Gamma_err": Gamma_err})
fits = pd.DataFrame(rows)
fits
````

**What you should see.** A table with nine rows. $\Gamma$ rises from about 3400 s⁻¹ at 30° to about 46 000 s⁻¹ at 150°. The value at 90° is the same as in Step 10.

**How to read it.** $\beta$ stays between 0.72 and 0.80 with no trend related to the rate. It is lower at the smallest angle, and that is a property of the optics. $\Gamma$ changes by more than a factor of 13.

**Common errors.** `KeyError: 'q_nm-1'` or `KeyError: 'angle_deg'`: check the spelling of the column name. `NameError: name 'g2_model' is not defined`: you skipped Step 8.

# Part E. Is it diffusion? And how big is the protein?

## The test

Diffusion makes one sharp prediction: $\Gamma = Dq^2$. Plot $\Gamma$ against $q^2$, and the points must fall on **a straight line through zero**.

Why this is a strong test: three things must hold together.

1. The points lie on a line, so the rate scales as $q^2$ and not as, for example, $q$ (which would be ballistic motion or flow).
2. The line goes through zero, because a ripple of infinite wavelength, $q\to0$, can't decay by diffusion. A rate left over at $q = 0$ would point to something that does not depend on length scale, such as a chemical process or a drift in the sample.
3. The slope is $D$, one number for all angles.

If the line curved, or missed zero, that would point to a second population, aggregates or interactions.

## Step 13. Plot $\Gamma$ against $q^2$

**What this cell does and why.** We make the plot and look at it before fitting a line.

- **`fits["q"]**2`** squares each value in the column, giving $q^2$ in nm⁻².
- **`plt.errorbar(x, y, yerr=..., fmt="o")`** plots points with error bars. Here the error bars are smaller than the markers.
- **`plt.xlim(left=0)`**, **`plt.ylim(bottom=0)`** make the axes start at zero, so that we can judge the line through zero by eye.

````python
x = fits["q"].values**2                # q squared, in nm^-2
y = fits["Gamma"].values               # decay rate, in 1/s
y_err = fits["Gamma_err"].values
plt.errorbar(x, y, yerr=y_err, fmt="o", label="fits at nine angles")
plt.xlim(left=0); plt.ylim(bottom=0)
plt.xlabel(r"$q^2$ (nm$^{-2}$)"); plt.ylabel(r"$\Gamma$ (s$^{-1}$)")
plt.legend(); plt.show()
````

**What you should see.** Nine points on what looks like a straight line, heading towards the origin. The points get closer together at large $q^2$, because $\sin^2(\theta/2)$ flattens out near 180°.

**What it means.** By eye the test passes. Now we quantify it.

## Step 14. Fit a straight line with a free intercept

**What this cell does and why.** We allow a straight line $\Gamma = \text{slope}\times q^2 + \text{intercept}$. If the motion is diffusion, the fitted intercept must be zero within its uncertainty.

**`np.polyfit(x, y, 1, w=1/y_err, cov=True)`** fits a polynomial:

| Argument | Meaning |
|---|---|
| `x`, `y` | the data |
| `1` | the degree of the polynomial; degree 1 is a straight line |
| `w=1/y_err` | weights. A point with a small error counts more. For `polyfit`, the weight is 1/error, not 1/error². |
| `cov=True` | also give back the covariance matrix, to get uncertainties |

**What comes back:** the coefficients, **highest power first**, so `(slope, intercept)`, and the covariance matrix. With `cov=True`, `polyfit` scales the uncertainties by the observed scatter of the points around the line. We will come back to this in the Bonus part.

````python
(slope, intercept), cov = np.polyfit(x, y, 1, w=1 / y_err, cov=True)
slope_err, intercept_err = np.sqrt(np.diag(cov))
print(f"slope     = {slope:.4g} +- {slope_err:.2g} nm^2/s")
print(f"intercept = {intercept:.0f} +- {intercept_err:.0f} per second")
print(f"intercept / its uncertainty = {intercept / intercept_err:.1f}")
````

**What you should see.** A slope of about $7.2\times10^{7}\ \mathrm{nm^2/s}$ and an intercept of about $23 \pm 25\ \mathrm{s^{-1}}$.

**What it means.** The intercept is about 1 uncertainty away from zero, and it is less than 0.1 % of the rate at 90°. **The line goes through zero. This is diffusion.** That answers the question left open by the lecture.

Having passed the test, we can now fix the intercept at zero, because the physics says it is zero, and get the best value of $D$.

## Step 15. Get $D$ from a line through zero

**What this cell does and why.** For $\Gamma = Dq^2$ with weights $w_i = 1/\sigma_i^2$, minimising $\chi^2$ gives a simple closed formula. You can derive it by setting $d\chi^2/dD = 0$:

$$D = \frac{\sum_i w_i\,q_i^2\,\Gamma_i}{\sum_i w_i\,q_i^4}, \qquad \sigma_D = \frac{1}{\sqrt{\sum_i w_i\,q_i^4}} .$$

**Units and conversions.** $\Gamma$ is in s⁻¹ and $q^2$ in nm⁻², so $D$ comes out in nm²/s. Two conversions:

- to μm²/s: 1 μm² = 10⁶ nm², so **divide by 10⁶**.
- to m²/s, the SI unit needed for the next step: 1 nm² = 10⁻¹⁸ m², so **multiply by 10⁻¹⁸**.

````python
w = 1 / y_err**2                                  # weights
D = np.sum(w * x * y) / np.sum(w * x * x)         # in nm^2/s
D_err = 1 / np.sqrt(np.sum(w * x * x))
print(f"D = {D / 1e6:.2f} +- {D_err / 1e6:.2f} um^2/s")
print(f"D = {D * 1e-18:.3e} m^2/s")
plt.errorbar(x, y, yerr=y_err, fmt="o", label="fits at nine angles")
plt.plot([0, x.max()], [0, D * x.max()], "-", label=r"$\Gamma = Dq^2$")
plt.xlim(left=0); plt.ylim(bottom=0)
plt.xlabel(r"$q^2$ (nm$^{-2}$)"); plt.ylabel(r"$\Gamma$ (s$^{-1}$)"); plt.legend(); plt.show()
````

**What you should see.** $D \approx 72.2\ \mathrm{\mu m^2/s}$, or $7.22\times10^{-11}\ \mathrm{m^2/s}$, and a line through the points and the origin.

`plt.plot([x1, x2], [y1, y2], "-")` draws a straight line between two points.

**Is the number sensible?** In one second a BSA molecule travels about $\sqrt{6Dt} \approx 21\ \mathrm{\mu m}$ in three dimensions. That is a typical value for a protein in water.

The uncertainty printed here is too small. Why, and by how much, is the subject of the Bonus part.

## Step 16. The hydrodynamic radius

**What this cell does and why.** Stokes–Einstein balances the two sides of the same motion. A density fluctuation pushes particles outward by osmotic pressure ($k_BT$ per particle), and the liquid resists with Stokes drag ($6\pi\eta R$). So

$$R_h = \frac{k_BT}{6\pi\eta D}.$$

$R_h$ is **defined** by this equation: it is the radius of the sphere with the same $D$ as our protein.

**Check the units first.** $k_BT$ is in J = kg·m²/s². $\eta$ is in Pa·s = kg/(m·s). $D$ is in m²/s. So

$$\frac{\mathrm{kg\,m^2/s^2}}{\mathrm{kg/(m\,s)}\cdot\mathrm{m^2/s}} = \mathrm{m}.$$

The result comes out in metres, so we multiply by $10^9$ to get nm. The relative uncertainty of $R_h$ equals that of $D$, because $R_h\propto 1/D$.

````python
D_si = D * 1e-18                                   # nm^2/s to m^2/s
R_h = KB * T / (6 * np.pi * ETA * D_si) * 1e9      # m to nm
R_h_err = R_h * D_err / D
print(f"k_B T = {KB * T:.3e} J")
print(f"R_h = {R_h:.2f} +- {R_h_err:.3f} nm")
````

**What you should see.** $k_BT \approx 4.12\times10^{-21}\ \mathrm{J}$ and $R_h \approx 3.40\ \mathrm{nm}$.

**How does this compare with the literature?** Guidolin et al. (*Macromolecules* **56**, 8290, 2023) report $R_h = 3.4 \pm 0.1$ nm for BSA from DLS, extrapolated to zero concentration, and list published values between 3.3 and 4.3 nm. Our value lies in that range.

**This also explains the lecture.** The lecture's rough guess of $r \approx 3$ nm predicted $\Gamma \approx 28\,500\ \mathrm{s^{-1}}$ at 90°. The observed rate is lower by a factor of about 3.4/3, and now you can see why.

**Common errors**

| Symptom | Cause | Fix |
|---|---|---|
| $R_h \approx 3\times10^{-9}$ | The factor `1e9` is missing; the result is in metres. | Keep the conversion. |
| $R_h \approx 3400$ or $0.0034$ | $\eta$ was entered in mPa·s, not Pa·s, or $D$ was not converted to m²/s. | Use SI inside the formula. |
| $R_h \approx 1.7$ nm | You fitted without the 2 in the exponent. | See Question 4. |

# Part F. Three radii for one protein: DLS against SAXS

## Why the radii differ: the key physics

A protein is not a sphere, so "its radius" is not one number. Each technique defines the radius by what it responds to.

| Radius | Defined by | Responds to |
|---|---|---|
| $R_g$, radius of gyration | SAXS, from the low-$q$ decay of $I(q)$ | how the **mass** (more exactly the excess electron density) is spread around the centre: $R_g^2$ is its mean squared distance from the centre |
| $R_V$, volume-equivalent radius | geometry | the radius of the sphere with the same **volume** |
| $R_h$, hydrodynamic radius | DLS, through Stokes–Einstein | **friction** against the solvent: it includes the shape, and the water that moves along with the protein |

They would differ even for a perfect sphere. For a solid sphere of radius $R$, integrating over the volume gives $R_g^2 = \tfrac{3}{5}R^2$, so

$$\frac{R_g}{R_h} = \sqrt{\tfrac{3}{5}} = 0.775 \quad \text{(solid sphere).}$$

So the question is not "which is right?", because all are right. The question is what the differences teach us.

## The SAXS model of BSA

Zhang et al. fitted dilute BSA in SAXS with the form factor of an **oblate ellipsoid**, with semi-axes $17 \times 42 \times 42$ Å, which are $1.7 \times 4.2 \times 4.2$ nm, and with an uncertainty of about ±1 Å on each.

*F. Zhang, M. W. A. Skoda, R. M. J. Jacobs, R. A. Martin, C. M. Martin, F. Schreiber, "Protein interactions studied by SAXS: effect of ionic strength and protein concentration for BSA in aqueous solutions", J. Phys. Chem. B **111**, 251–259 (2007).*

An ellipsoid with semi-axes $a, b, c$ and uniform density has

$$R_g = \sqrt{\frac{a^2+b^2+c^2}{5}}, \qquad R_V = (abc)^{1/3}.$$

Check: for $a=b=c=R$, these give $R_g = \sqrt{3/5}\,R$ and $R_V=R$, as they should.

## Step 17. Calculate $R_g$ and $R_V$ from the axes

**What this cell does and why.** We compute both radii, and then compare them with $R_h$. Dividing Å by 10 gives nm. `1/3` as a power gives a cube root, so `x**(1/3)` is $x^{1/3}$.

````python
a, b, c = 1.7, 4.2, 4.2                    # <-- CHANGE ME   ellipsoid semi-axes in nm
R_g = np.sqrt((a**2 + b**2 + c**2) / 5)    # radius of gyration, nm
R_V = (a * b * c)**(1 / 3)                 # volume-equivalent radius, nm
print(f"R_g from the axes = {R_g:.2f} nm")
print(f"R_V from the axes = {R_V:.2f} nm")
print(f"R_h from our fit  = {R_h:.2f} nm")
````

**What you should see.** $R_g \approx 2.76$ nm (2.8 nm when rounded; Zhang et al. quote 27.6 Å for this shape), $R_V \approx 3.11$ nm, and $R_h \approx 3.40$ nm.

**Directly from SAXS data.** The value of $R_g$ can also be obtained without any shape model, by a Guinier analysis of the low-$q$ scattering. Trewhella et al. report for BSA $R_g = 28.33 \pm 0.05$ Å from a Guinier analysis (Table 5 of *Acta Cryst. D* **73**, 710–728, 2017). That is 2.83 nm, close to the value from the ellipsoid.

So: $R_g < R_V < R_h$. Let us understand this order.

## Step 18. Compare $R_g/R_h$ with the solid sphere

**What this cell does and why.** The ratio $R_g/R_h$ removes the overall size and leaves information about shape. `np.sqrt(3 / 5)` is the solid-sphere value.

````python
R_g_direct = 2.833                           # <-- CHANGE ME   nm, Guinier value, Trewhella et al. 2017
print(f"solid sphere              : {np.sqrt(3 / 5):.3f}")
print(f"R_g(ellipsoid) / R_h      : {R_g / R_h:.3f}")
print(f"R_g(Guinier)   / R_h      : {R_g_direct / R_h:.3f}")
````

**What you should see.** 0.775 for the sphere, 0.813 with the ellipsoid's $R_g$, and 0.834 with the Guinier $R_g$.

**How to read it.** The protein's ratio is above that of a solid sphere. That is what a non-spherical shape does: pushing mass away from the centre raises $R_g$ quickly. Friction responds more weakly, as we now calculate.

## Step 19. The Perrin shape correction

**What this cell does and why.** How much friction does the flat shape add? Perrin calculated this for ellipsoids: the friction of an ellipsoid is that of the sphere with the same volume, multiplied by a factor $F \ge 1$. So the radius predicted from shape alone is

$$R_P = F\,R_V .$$

For an oblate ellipsoid with axial ratio $p = a/b < 1$ (polar semi-axis over equatorial semi-axis),

$$F = p^{2/3}\,\frac{\xi}{\arctan\xi}, \qquad \xi = \frac{\sqrt{1-p^2}}{p}.$$

(C. R. Cantor, P. R. Schimmel, *Biophysical Chemistry, Part II*, Freeman, 1980.)

**New commands.** **`np.arctan`** is the inverse tangent, in radians. Check of the limit: for $p \to 1$, $\xi\to0$ and $\arctan\xi\to\xi$, so $F\to1$, the sphere.

````python
p = a / b                                    # axial ratio, oblate: p < 1
xi = np.sqrt(1 - p**2) / p
F = p**(2 / 3) * xi / np.arctan(xi)          # Perrin friction factor
R_P = F * R_V                                # radius predicted from shape alone
gap = R_h - R_P
print(f"axial ratio p = {p:.3f}, Perrin factor F = {F:.3f}")
print(f"R_V = {R_V:.2f} nm  -->  R_P = F R_V = {R_P:.2f} nm,   R_h = {R_h:.2f} nm")
print(f"remaining gap = {gap:.2f} nm, which is {100 * gap / R_h:.0f} % of R_h")
````

**What you should see.** $p = 0.405$, $F = 1.071$, $R_P = 3.33$ nm, and a remaining gap of about 0.07 nm, or 2 %.

**How to read it.** A flattening of 2.5 to 1 adds only 7 % to the friction. That is the weak response mentioned above: $R_h$ hardly notices the shape, compared with the change in $R_g$. Shape takes us from 3.11 nm most of the way to 3.40 nm.

## Step 20. How much of the gap could hydration explain?

**What this cell does and why.** Water close to the protein surface moves with the protein, and so adds to its friction. How does that compare with the gap? We try two reference points, and they teach different things.

**Reference 1: the SAXS ellipsoid.** A water molecule is about 0.3 nm across. What would one layer of water around the ellipsoid do? We add it to $R_V$ and multiply by $F$.

**Reference 2: the dry protein.** The volume of the bare protein follows from its mass, with the typical partial specific volume $0.73\ \mathrm{cm^3/g}$:

$$V_\mathrm{dry}\,[\mathrm{nm^3}] = 1.212\times10^{-3}\ M\,[\mathrm{Da}]$$

(H. P. Erickson, *Biol. Proced. Online* **11**, 32–51, 2009). BSA has $M \approx 66$ kDa (Zhang et al.). The sphere with this volume has radius $R_\mathrm{dry} = (3V/4\pi)^{1/3}$, the smallest sphere that can hold the protein.

````python
R_one_layer = F * (R_V + 0.3)                       # ellipsoid plus one layer of water, nm
V_dry = 1.212e-3 * 66000                            # nm^3, from the mass
R_dry = (3 * V_dry / (4 * np.pi))**(1 / 3)          # nm
V_ell = 4 / 3 * np.pi * a * b * c                   # volume of the SAXS ellipsoid, nm^3
print(f"Ellipsoid plus one water layer: {R_one_layer:.2f} nm   (R_h = {R_h:.2f} nm)")
print(f"Dry volume {V_dry:.0f} nm^3, R_dry = {R_dry:.2f} nm")
print(f"Ellipsoid volume {V_ell:.0f} nm^3, which is {V_ell / V_dry:.2f} times the dry volume")
print(f"Dry protein with shape: F R_dry = {F * R_dry:.2f} nm,  gap to R_h = {R_h - F * R_dry:.2f} nm")
````

**What you should see.** An ellipsoid with one water layer would have 3.65 nm, more than $R_h$. The dry volume is 80 nm³ with $R_\mathrm{dry} = 2.67$ nm. The ellipsoid's volume is 126 nm³, or 1.57 times larger. And $F R_\mathrm{dry} = 2.86$ nm, which leaves a gap of 0.53 nm.

**How to read it: a surprise.**

- Compared with the **SAXS ellipsoid**, the gap of 0.07 nm is much smaller than one layer of water. Hydration more than covers it. A full extra layer would overshoot.
- Yet the same ellipsoid already holds **57 % more volume than the dry protein**. A uniform ellipsoid fitted to SAXS is a smooth envelope around a rough molecule, and it includes surface grooves and part of the hydration water, which also scatter X-rays. So the ellipsoid has already swallowed much of what we hoped to attribute to "hydration".
- Compared with the **dry protein**, the gap is 0.53 nm, roughly two layers of water. Water, plus the roughness of the surface, can account for it.

**The conclusion, stated honestly.** We can't read off "the" amount of bound water from these numbers. With an ellipsoid accurate to ±0.1 nm on each semi-axis, and a published range of $R_h$ from 3.3 to 4.3 nm, a gap of 0.07 nm is not significant. What the comparison does show is the order of the effects: $R_V$ to $R_h$ is explained by a modest shape correction plus a thin, partly hidden layer of water. It is not a large effect, and the three radii are mutually consistent.

**Link to Tuesday's SAXS lecture.** $R_g$ comes from how $I(q)$ falls at small $q$: it is a static property that weights the **mass distribution**, and it does not care about solvent flow. $R_h$ comes from how fast the speckle decays: it is a dynamic property that weights **friction**. The same molecule, seen by two different questions, gives two different and equally correct lengths.

# Bonus. How honest are my error bars?

*For those with time, or with Python experience.*

## The hint from the lecture

**Neighbouring lags in a correlator share noise, so the error bars of a simple fit are too small.**

Why? `curve_fit` assumes that the error of each point is independent of all others. Then 136 points would give about $\sqrt{136}$ times more precision than one point. But in a correlator, neighbouring lags are built from the same photons in the same stretch of time. If one point is high by chance, its neighbours tend to be high as well. You saw this as slow waves at long lags in the residuals of Step 11. So there are fewer independent points than it seems, and the true uncertainty is larger. Worse, if noise moves the whole decay a little to the left or right, the fit simply follows it: $\Gamma$ is off, and the residuals show nothing.

## The test: scatter about the line

We need an independent check of the error bars, and we have one. Theory says nine rates lie on one line, so **the scatter of the nine points about the line measures their true error**.

Compute the reduced $\chi^2$ of the line. If the errors were honest it would be near 1. If it is much larger, the errors are too small by a factor of about $\sqrt{\chi^2_\mathrm{red}}$, and we scale them by that factor.

````python
chi2_line = np.sum(w * (y - D * x)**2) / (len(x) - 1)     # 9 points, 1 parameter
scale = np.sqrt(chi2_line)
print(f"reduced chi-squared about the line = {chi2_line:.1f},  so scale the errors by {scale:.1f}")
print(f"D   = {D / 1e6:.2f} +- {scale * D_err / 1e6:.2f} um^2/s   (was +- {D_err / 1e6:.2f})")
print(f"R_h = {R_h:.2f} +- {scale * R_h_err:.3f} nm   (was +- {R_h_err:.3f})")
````

**What you should see.** A reduced $\chi^2$ of about 10.6, so errors too small by a factor of about 3.3. Then $D = 72.21 \pm 0.15\ \mathrm{\mu m^2/s}$ and $R_h = 3.40 \pm 0.007$ nm.

**How to read it.** The fits at each angle looked fine (reduced $\chi^2$ near 1, because each point's error hides the correlation), yet the fitted rates scatter about three times more than their stated errors. This is the same scaling that `np.polyfit` applied silently in Step 14 with `cov=True`. That is why the intercept's error was already realistic there. With `cov="unscaled"`, you get the small error back.

**A warning.** Even $\pm 0.007$ nm is a statistical error only. The viscosity of the buffer, or the temperature, could shift $R_h$ by a few percent, a far larger effect. Don't quote a radius of a protein to 0.2 %.
