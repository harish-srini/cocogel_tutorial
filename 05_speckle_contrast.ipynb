## What you will learn

- That a speckle pattern comes from nothing but random phases.
- That its intensity has an exponential distribution, which gives $\beta = 1$ for one speckle.
- Why $\beta$ falls as a detector collects more speckles, and that $\beta$ sets the height of the decay and not its rate.
- The trade-off between signal and contrast, and how XPCS deals with it.

## The physics, in words

Each particle in the beam sends out a wave with its own phase, because each sits at a random position. On a far-away screen, all those waves add up. At some points they happen to add up, and at others they cancel. That is the speckle pattern: bright and dark spots, with no lens and no structure in the sample needed.

The spots have a typical size, set by geometry: the larger the illuminated region, the smaller the speckles.

A detector that is smaller than a speckle sees one speckle. As the particles move, it sees the full flicker between dark and bright, and $\beta = 1$. A detector that covers many speckles adds up many independent flickers. When one goes up, another goes down, and the sum flickers less. The flicker is still there and has the same speed, but its size relative to the mean is smaller. That size is $\beta$.

## The physics, in equations

**The field at one point** is a sum of many waves with random phases:

$$E = \sum_j e^{i\phi_j}.$$

By the central limit theorem, $E$ is a complex Gaussian variable. Then the intensity $I = |E|^2$ has an **exponential distribution**,

$$P(I) = \frac{1}{\langle I\rangle}e^{-I/\langle I\rangle},$$

so the darkest intensities are the most likely, and very bright spots are rare. For this distribution the variance equals the mean squared:

$$\frac{\langle I^2\rangle - \langle I\rangle^2}{\langle I\rangle^2} = 1 .$$

**The link to $g_2$.** At $\tau\to0$ the intensity hasn't changed, so $g_2(0) = \langle I^2\rangle/\langle I\rangle^2$. With $f(q,0)=1$, $g_2(0) = 1+\beta$. Therefore

$$\beta = \frac{\langle I^2\rangle - \langle I\rangle^2}{\langle I\rangle^2},$$

measured on what the detector actually records. **So we can find $\beta$ from a snapshot.**

**Collecting $M$ independent speckles.** The mean of $M$ independent values has a variance $M$ times smaller, so for large $M$

$$\beta \approx \frac{1}{M}, \qquad M = \frac{A_\mathrm{detector}}{A_\mathrm{speckle}} .$$

**What sets the speckle size.** For a wavelength $\lambda$, an illuminated region of size $a$ and a distance $L$ to the detector, the speckle size $s$ is about

$$s \approx \frac{\lambda L}{a}.$$

# Part A. A speckle pattern, from nothing but random phases

## Step 4. Build the pattern

**What this cell does and why.** Each point of the illuminated region sends out a wave with a random phase. In the far field, the total field is the **Fourier transform** of the field in the illuminated region, so one command does the whole interference sum.

| Code | What it does |
|---|---|
| `np.random.default_rng(seed)` | Creates a source of random numbers, starting from `seed`. |
| `rng.random((n, n))` | An `n` × `n` array of random numbers between 0 and 1. |
| `np.indices((n, n))` | Two arrays holding the row number and the column number of each pixel. Subtracting `n // 2` moves the origin to the centre. |
| `y**2 + x**2 < (d / 2)**2` | `True` inside a circle of diameter `d` and `False` outside. In arithmetic, `True` counts as 1 and `False` as 0. |
| `np.exp(2j * np.pi * rng.random(...))` | A random phase $e^{i\phi}$ with $\phi$ between 0 and $2\pi$, at each pixel. `np.pi` is the number π. |
| `np.fft.fft2(a)` | The two-dimensional fast Fourier transform: the far field. |
| `np.abs(E)**2` | The intensity $I = |E|^2$. |

We divide by the mean, **`.mean()`**, so that $\langle I\rangle = 1$. **`.sum()`** adds up all elements, and here it counts the illuminated pixels.

````python
def make_speckle(lit_px, seed=SEED, n=N):
    rng = np.random.default_rng(seed)
    y, x = np.indices((n, n)) - n // 2
    lit = y**2 + x**2 < (lit_px / 2)**2                         # the illuminated circle
    E = np.fft.fft2(lit * np.exp(2j * np.pi * rng.random((n, n))))  # all waves added up
    I = np.abs(E)**2
    return I / I.mean(), n**2 / lit.sum()                       # intensity, and speckle area
I, A_speckle = make_speckle(LIT_PX)
print(f"Speckle area: {A_speckle:.0f} pixels, so about {np.sqrt(A_speckle):.0f} pixels across")
````

**What you should see.** A speckle area of about 326 pixels, so a speckle is about 18 pixels across.

**Where does the speckle area come from?** It is the area of the pattern divided by the number of illuminated pixels, $N^2/N_\mathrm{lit}$. The formula follows from the properties of the Fourier transform: illuminating more pixels gives more independent speckles in the same plane.

**Common error.** `MemoryError` on a small laptop: set `N = 1024` in Step 3.

## Step 5. Look at it, and test the exponential law

**What this cell does and why.** We show a piece of the pattern, and a histogram of the intensities. The prediction is $P(I) = e^{-I}$, with no free parameter at all.

| Code | What it does |
|---|---|
| `plt.subplots(1, 2, figsize=(12, 4.8))` | A figure with two panels side by side. |
| `a.imshow(image, cmap="gray", vmax=5)` | Shows an array as an image. `cmap` is the colour scale, and `vmax` the value that appears white. |
| `b.hist(values, bins=100, density=True, range=(0, 8))` | A histogram, normalised so that its area is 1. `values` must be a flat list, so `.ravel()` turns the 2D array into a 1D one. |
| `b.set_yscale("log")` | A logarithmic y axis, on which an exponential is a straight line. |
| `np.linspace(a, b, n)` | `n` evenly spaced numbers from `a` to `b`. |
| `.var()` | The variance. |

````python
fig, (a, b) = plt.subplots(1, 2, figsize=(12, 4.8))
a.imshow(I[:256, :256], cmap="gray", vmax=5); a.set_title("a piece of the pattern")
b.hist(I.ravel(), bins=100, density=True, range=(0, 8), label="pattern")
i_values = np.linspace(0, 8, 100); b.plot(i_values, np.exp(-i_values), "k--", label=r"$e^{-I}$")
b.set_yscale("log"); b.set_xlabel(r"intensity $I/\langle I\rangle$"); b.set_ylabel("probability density")
b.legend(); plt.show()
print(f"beta of the pattern = {I.var() / I.mean()**2:.3f}")
````

**What you should see.** Left: a pattern of bright grains on a dark background. It looks like laser light on a wall. Right: a histogram that follows the dashed straight line. And $\beta \approx 1.01$, within the statistics of 1.

**How to read it.** We put in nothing but random phases, and the statistics come out by themselves. On the left, most of the area is dark and a few spots are very bright, which is what the exponential law says. **One speckle has contrast 1: the largest possible.**

**Question 1.** What fraction of the pixels is brighter than three times the mean? Answer from the formula, then check it with `np.mean(I > 3)`.

The integral of e^(−I) from 3 to infinity is e⁻³ ≈ 0.05, so 5 %. `np.mean(I > 3)` gives about the same, since the mean of `True`/`False` values is the fraction that are `True`.

# Part B. A detector averages over speckles

## Step 6. Change the size of the detector

**What this cell does and why.** A detector records the mean intensity over its area. So we cut the pattern into square blocks of side `detector_px`, and take the mean of each. Each block is one possible detector position. Then $\beta$ is the variance of those values over the mean squared.

**How the blocks are made.** `I[:k*m, :k*m]` trims the array to a whole number of blocks. **`.reshape(k, m, k, m)`** arranges it as blocks of `m` × `m`, and **`.mean(axis=(1, 3))`** takes the mean within each block.

**Try:** `1`, `4`, `16`, `64`. After each change, press Shift+Enter on the second of these two cells.

````python
def detector_values(I, m):
    k = I.shape[0] // m                                   # number of blocks per side
    return I[:k * m, :k * m].reshape(k, m, k, m).mean(axis=(1, 3))
def beta_of(values):
    return values.var() / values.mean()**2
````

````python
detector_px = 16                    # <-- CHANGE ME   side of the detector, in pixels
seen = detector_values(I, detector_px); M = detector_px**2 / A_speckle
fig, (a, b) = plt.subplots(1, 2, figsize=(12, 4.8))
a.imshow(I[:256, :256], cmap="gray", vmax=5, extent=(0, 256, 256, 0)); a.set_title("the pattern")
k = 256 // detector_px              # the same area, seen through detectors of this size
b.imshow(seen[:k, :k], cmap="gray", vmin=0, vmax=5, extent=(0, 256, 256, 0))
b.set_title("what detectors record"); plt.show()
print(f"Detector {detector_px} pixels:  speckles collected M = {M:.2f},  beta = {beta_of(seen):.3f}")
````

**What you should see**

| `detector_px` | speckles collected, $M$ | $\beta$ | the right-hand image |
|---|---|---|---|
| 1 | 0.003 | 1.01 | the same as the pattern |
| 4 | 0.05 | 0.96 | almost the same |
| 16 | 0.78 | 0.51 | blocks, still with strong differences |
| 64 | 12.5 | 0.06 | nearly uniform grey |

**How to read it.** The right-hand image shows the same area, with the same grey scale. As the detector grows, the differences between one detector position and another shrink. Remember: for moving particles, the speckle pattern changes in time at a fixed detector, so these same differences appear as flicker. **Small differences mean a small flicker, and a small $\beta$.**

`extent` tells `imshow` which coordinates the image covers, so that both panels show the same area.

**Common error.** `ValueError: cannot reshape array` can occur if `detector_px` is larger than `N`. Use a smaller value.

# Part C. $\beta$ against the number of speckles

## Step 7. The curve

**What this cell does and why.** We run through many detector sizes. **`plt.loglog`** gives logarithmic axes on both sides, on which a power law is a straight line. We compare with $\beta = 1/M$, and with a rule of thumb that works at all sizes, $\beta \approx 1/(1+M)$.

````python
sizes = np.array([1, 2, 4, 8, 16, 32, 64, 128])
betas = np.array([beta_of(detector_values(I, m)) for m in sizes])
M_values = sizes**2 / A_speckle
plt.loglog(M_values, betas, "o", label="from the pattern")
plt.loglog(M_values, 1 / (1 + M_values), "-", label=r"$1/(1+M)$")
plt.loglog(M_values[4:], 1 / M_values[4:], "k--", label=r"$1/M$")
plt.xlabel("speckles collected, $M$"); plt.ylabel(r"contrast $\beta$"); plt.legend(); plt.show()
print(np.round(betas, 3))
````

**What you should see.** Points at $\beta\approx1$ for small detectors, falling along a line of slope −1 on the right. The solid curve is a good guide at small $M$. At large $M$ the points lie below both curves, by up to about a third, so treat it as a rule of thumb.

`[... for m in sizes]` is a *list comprehension*: a short way to build a list with a loop inside.

**So the rule is simple.** $\beta$ is 1 for one speckle or less, and about $1/M$ for many.

**Question 2.** You double the side of a detector that already covers many speckles. What happens to $\beta$, and what happens to the mean count rate?

<details>
<summary>Answer</summary>

The area grows by 4, so β falls by 4, and the count rate rises by 4. That is the trade-off: more signal, less contrast. Since the noise in g₂ depends on both, there's a best size, usually of the order of one speckle for a single detector.

</details>

# Part D. What sets the size of a speckle?

## Step 8. Change the illuminated region

**What this cell does and why.** The prediction is $s\approx\lambda L/a$: **a larger illuminated region gives smaller speckles.** In our calculation, $a$ is `LIT_PX`, and we can test the prediction with no change to anything else.

````python
lit_values = [64, 128, 256]                       # <-- CHANGE ME   diameters of the illuminated region
fig, axes = plt.subplots(1, len(lit_values), figsize=(13, 4.5))
for ax, lit_px in zip(axes, lit_values):          # zip pairs up the panels and the values
    I_d, A_d = make_speckle(lit_px)
    ax.imshow(I_d[:256, :256], cmap="gray", vmax=5)
    ax.set_title(f"lit {lit_px} px, speckle {np.sqrt(A_d):.0f} px")
plt.show()
````

**What you should see.** Three patterns, with speckles about 36, 18 and 9 pixels across. Double the illuminated size, and the speckle size halves.

**With round numbers, for illustration only:**

| | $\lambda$ | $a$ | $L$ | speckle size $\lambda L/a$ |
|---|---|---|---|---|
| Visible light | 600 nm | 100 μm | 30 cm | about 2 mm |
| Hard X-rays | 0.1 nm | 10 μm | 5 m | 50 μm |

So in visible light, a speckle is a millimetre or two across, and a pinhole or a single-mode fibre can easily select one or a few. That is why $\beta$ of order 0.8 is within reach in DLS. For X-rays, a speckle is about as large as a detector pixel or smaller, since pixels are typically tens of μm to a few hundred μm. So $\beta$ falls fast, and a larger $s$ means a long distance or a very small beam.

**How XPCS turns this around.** A pixel array holds many pixels, each seeing about one speckle or more. So one computes the correlation at each pixel and averages over the pixels of a ring of equal $q$. This keeps the contrast of one pixel while gaining the statistics of many. It is the multi-speckle idea, also used with light and a camera (Cipelletti et al., *Phys. Rev. Lett.* **84**, 2275, 2000).

**The contrast also falls for reasons other than geometry.** X-ray beams are only partly coherent, and the stray light or a poor optical alignment reduces the contrast in DLS. So the "effective" number of speckles can be larger than the geometric count.

**The open question:** a low contrast sounds like bad news. Does it also change the rate?

# Part E. Does $\beta$ change the rate?

## Step 9. A moving speckle

**What this cell does and why.** Now we let time run. Each speckle's field $E_m(t)$ changes step by step, and loses its memory at a rate $\Gamma$:

$$E(t+1) = a\,E(t) + \sqrt{1-a^2}\;\xi(t), \qquad a = e^{-\Gamma}.$$

Here $\xi$ is a new complex Gaussian random number at each step. This has $f(\tau) = e^{-\Gamma\tau}$, with $\tau$ in frames, and the field stays Gaussian at all times, just as a field from many particles does. We choose $1/(2\Gamma) = 15$ frames, as in the lecture's speckle movies.

The detector records the mean intensity of $M$ independent speckles, at each time.

| Code | What it does |
|---|---|
| `rng.normal(size=(T, M))` | A `T` × `M` array of Gaussian random numbers. |
| `z = a * z + c * noise[t]` | One step for all `M` speckles at once. |
| `(np.abs(E)**2).mean(axis=1)` | Mean over the speckles, for each time. `axis=1` means along the columns. |

````python
def detector_trace(M, Gamma, T=200000, seed=SEED):
    rng = np.random.default_rng(seed); a, c = np.exp(-Gamma), np.sqrt(1 - np.exp(-2 * Gamma))
    noise = (rng.normal(size=(T, M)) + 1j * rng.normal(size=(T, M))) / np.sqrt(2)
    z, E = noise[0], np.zeros((T, M), dtype=complex)
    for t in range(T):                                 # step through time
        z = a * z + c * noise[t]; E[t] = z
    return (np.abs(E)**2).mean(axis=1)                 # what the detector records
````

**What you should see.** Nothing is printed. This cell only defines the function.

**`range(T)`** gives the whole numbers from 0 to `T` − 1. **`np.zeros`** makes an array filled with zeros, and `dtype=complex` allows complex numbers.

## Step 10. Calculate $g_2$, for different numbers of speckles

**What this cell does and why.** This is the definition of $g_2$ from the lecture: $\langle I(t)I(t+\tau)\rangle/\langle I\rangle^2$. `I_t[:-lag]` is all values but the last `lag`, and `I_t[lag:]` is all but the first `lag`, so multiplying them pairs each value with the one `lag` frames later. We fit it with the same model as in notebook 01.

**`curve_fit`** adjusts the parameters of a model to match the data. `p0` is the starting guess. It gives back the best parameters and a covariance matrix, which we ignore here with `_`.

This cell takes a few seconds. **Try** other values in `speckle_counts`.

````python
GAMMA = 1 / 30                                        # per frame
lags = np.array([1, 2, 3, 5, 8, 12, 18, 27, 40, 60, 90, 135, 200, 300])    # frames
speckle_counts = [1, 4, 16, 64]                       # <-- CHANGE ME   speckles on the detector
for M in speckle_counts:
    I_t = detector_trace(M, GAMMA)
    g2 = np.array([np.mean(I_t[:-lag] * I_t[lag:]) for lag in lags]) / np.mean(I_t)**2
    (b_fit, G_fit), _ = curve_fit(lambda t, b, G: 1 + b * np.exp(-2 * G * t), lags, g2, p0=[1 / M, 0.03])
    plt.semilogx(lags, (g2 - 1) / b_fit, "o-", ms=4,
                 label=f"M = {M}, beta = {b_fit:.3f}, Gamma = {G_fit:.4f}")
plt.xlabel(r"lag $\tau$ (frames)"); plt.ylabel(r"$(g_2-1)/\beta$"); plt.legend(loc="lower left", fontsize=11); plt.show()
````

**What you should see.** Four curves that fall on top of each other, once they are divided by $\beta$. The legend shows $\beta$ close to 1, 0.25, 0.06 and 0.015, which is about $1/M$. It also shows $\Gamma$ within about 5 % of the true $1/30 = 0.0333$ per frame in all cases.

**How to read it: this answers the question.** **$\beta$ sets the height of the decay, and not its rate.** A low contrast doesn't bias $\Gamma$, $D$ or $R_h$. It only makes them harder to measure, since a smaller signal sits on top of the noise.

This also confirms the Siegert relation from the lecture, without assuming it: a Gaussian field gave us $g_2-1 = \beta e^{-2\Gamma\tau}$, with its factor of 2.

`lambda t, b, G: ...` is a one-line function without a name. A *lambda* works here just like a function written with `def`.
